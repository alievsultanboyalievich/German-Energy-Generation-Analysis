```python
import pandas as pd
```

```python
import sqlite3
```

```python
df = pd.read_csv('Actual_generation_012026102026_Hour.csv', encoding='Windows-1252', sep=';')
print(df.info())
print("\nMissing values per column:\n", df.isnull().sum())
df.head()
```

```python
df.column = (df.columns.str.strip()
                       .str.lower()
                       .str.replace(' ', '_')
                       .str.replace('[^a-z0-9_]', '', regex=True))

print(df.column.tolist())
```

```python
df.columns = (df.columns.str.strip()
                        .str.lower()
                        .str.replace(' ', '_')
                        .str.replace('[^a-z0-9_]', '', regex=True))

print(df.columns.tolist())
```

```python
df.head()
```

```python
df.columns = df.column.str.replace('calculated_resolutions', '')
df.head()
```

```python
df['start_date'] = pd.to_datetime(df['start_date'], format="mixed")
print(df['start_date'].dtype)

df['end_date'] = pd.to_datetime(df['end_date'], format="mixed")
print(df['end_date'].dtype)
df.head()
```

```python
energy_cols = [col for col in df.columns if col not in ['start_date', 'end_date']]
for col in energy_cols:
    df[col] = df[col].astype(str).str.replace(',', '').str.replace('-', '0')
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)
```

```python
df.head()
```

```python
df.info()
```

```python
# Strip trailing underscores from column names
df.columns = df.columns.str.rstrip('_')

# Display updated column names
print(df.columns.tolist())
```

```python
# Strip trailing underscores from column names
df.columns = df.columns.str.strip('_')

# Display updated column names
print(df.columns.tolist())
```

```python
df.head()
```

```python
df.columns = df.column.str.replace('_calculated_resolutions', '')
df.head()
```

```python
import sqlite3
```

```python
conn = sqlite3.connect(':memory:')
df.to_sql('energy_grid', conn, index=False, if_exists='replace')

sql_query = """
    SELECT 
        ROUND(SUM(wind_onshore_mwh), 2) AS total_wind_onshore,
        ROUND(SUM(photovoltaics_mwh), 2) AS total_solar
    FROM energy_grid;
    """
```

```python
energy_totals = pd.read_sql(sql_query, conn)
```
